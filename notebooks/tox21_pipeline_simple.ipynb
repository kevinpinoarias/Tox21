# Tox21 Pipeline Walkthrough

This notebook describes what the pipeline does step by step.


## 1. Load data

- Read the CSV file containing molecules and assay labels
- Identify:
  - `smiles` → molecular representation
  - `mol_id` → identifier
  - remaining columns → assay labels

- Remove:
  - rows with missing SMILES
  - duplicate molecules (optional)


## 2. Featurization

- Convert each SMILES string into a numeric vector
- Use Morgan fingerprints:
  - fixed-length binary vector (e.g. 2048 bits)
  - encodes local molecular structure

Output:
- `X` → feature matrix (n_samples × n_bits)


## 3. Target preparation

- Extract assay columns as labels
- Each assay is a separate binary classification task
- Labels may contain missing values (NaN)


## 4. Train/test split

- Split data into:
  - training set
  - test set

- Same split used for all assays


## 5. Model training (Logistic Regression)

- Train one model per assay
- For each assay:
  - remove rows with missing labels
  - train Logistic Regression on remaining data

Output:
- dictionary of trained models (one per assay)


## 6. Prediction

- Use trained models to predict on test set
- For each assay:
  - predicted class (0 or 1)
  - predicted probability

Outputs:
- `y_pred` → predicted labels
- `y_proba` → predicted probabilities


## 7. Evaluation

For each assay:
- accuracy
- precision
- recall
- F1 score
- ROC-AUC
- PR-AUC

Also compute:
- macro averages → average across assays
- micro averages → aggregate across all predictions


## 8. Plots

Generate:
- ROC curves per assay
- Precision-Recall curves per assay
- bar charts for metrics (e.g. ROC-AUC, F1)
- class balance per assay


## 9. Error analysis

Build a table with:
- molecule (SMILES, ID)
- assay name
- true label
- predicted label
- predicted probability

From this:
- identify misclassified samples
- rank most confident errors
- summarize errors per assay
- separate false positives and false negatives


## 10. Outputs

The pipeline saves:

- metrics → CSV files
- plots → images
- predictions → CSV files
- error analysis → CSV tables

All results are written to the `outputs/` folder.
